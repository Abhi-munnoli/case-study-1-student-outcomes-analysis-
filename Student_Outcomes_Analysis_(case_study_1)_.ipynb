🏫 Scenario
The administration at EduPulse University is concerned about declining graduation rates. They have provided you with a dataset to identify why students are struggling and how attendance correlates with final grades.



```python
from google.colab import files
import pandas as pd

uploaded = files.upload()
df = pd.read_csv(next(iter(uploaded)))
display(df)
```

1️⃣ Calculate the Average Marks
Compute the average marks of all students in the dataset.

```python
import pandas as pd
import io
df = pd.read_csv(io.BytesIO(uploaded["student_performance_dataset.csv"]))
average_marks = df["Marks"].mean()
print(f"Average Marks of All Students: {average_marks:.2f}")
```

2️⃣ High Performers Analysis
Filter the dataset to determine how many students scored above 75 marks.

```python
high_performers = df[df["Marks"] > 75]
print(high_performers)
print(f"\nNumber of students scoring above 75: {len(high_performers)}")
```

3️⃣ Correlation Analysis
Analyze whether higher attendance leads to better marks.
Calculate the correlation coefficient between Attendance and Marks.
Briefly interpret the result.

```python
import matplotlib.pyplot as plt
import seaborn as sns

display(df)
print("Correlation:", df["Attendance_Percentage"].corr(df["Marks"]))
sns.regplot(data=df, x="Attendance_Percentage", y="Marks")
plt.title("Attendance vs Marks")
plt.show()
```

4️⃣ Feature Engineering
Create a new column called Pass where:
Pass = True  if Marks >= 50  
Pass = False if Marks < 50

```python
df["Pass"] = df["Marks"] >= 50
display(df)
print(df["Pass"].value_counts())
```

5️⃣ Summary Statistics
Count the total number of:

✅ Passes
❌ Fails

```python
display(df)
print("Pass:", (df["Marks"] >= 50).sum())
print("Fail:", (df["Marks"] < 50).sum())
```

6️⃣ Visualization
Create a distribution plot (histogram) of student marks.

Add appropriate:

Title
Axis labels
Grid (optional)


```python
import pandas as pd
import matplotlib.pyplot as plt

display(df)

plt.hist(df["Marks"], bins=8, edgecolor="black")
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.grid()
plt.show()
```
